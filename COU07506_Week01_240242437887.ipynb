```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)
```

```python
## Task 1.1 — Your first notebook
```

```python
n = np.arange(-5, 11)

delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = (0.8 ** n) * step

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))
for ax, x, title in zip(axes, [delta, step, ramp, expo],
                        ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n (samples)")
    ax.set_ylabel("amplitude")

plt.tight_layout()
plt.show()
```

```python
## Task 1.2 — Change the parameters
```

```python
n = np.arange(-5, 16)
u = (n >= 0).astype(float)

fig, axes = plt.subplots(1, 3, figsize=(13, 3))
for ax, a in zip(axes, [0.5, -0.8, 1.1]):
    x = (a ** n) * u
    ax.stem(n, x, basefmt="k-")
    ax.set_title(f"({a})^n u(n)")
    ax.set_xlabel("n (samples)")
    ax.set_ylabel("amplitude")

plt.tight_layout()
plt.show()

n2 = np.arange(0, 41)
c1 = np.cos(2 * np.pi * n2 / 8)
c2 = np.cos(2 * np.pi * n2 / 7.5)

fig, axes = plt.subplots(2, 1, figsize=(10, 5), sharex=True)
axes[0].stem(n2, c1, basefmt="k-")
axes[0].set_title("cos(2πn / 8)")
axes[0].set_ylabel("amplitude")

axes[1].stem(n2, c2, basefmt="k-")
axes[1].set_title("cos(2πn / 7.5)")
axes[1].set_xlabel("n (samples)")
axes[1].set_ylabel("amplitude")

plt.tight_layout()
plt.show()
```

```python
**Observation:**
- For $a = 0.5$, $(a)^n u(n)$ decays smoothly toward 0. For $a = -0.8$, it decays while alternating sign. For $a = 1.1$, it grows exponentially without bound.
- $\cos(2\pi n / 8)$ repeats identically every **8 integer samples** (one full cycle). $\cos(2\pi n / 7.5)$ cannot repeat after 7.5 samples because $n$ must be an integer; since $7.5 = 15/2$ is rational, it repeats after **15 samples** (2 cycles).
```

```python
## Task 1.3 — Working with audio
```

```python
import os
from IPython.display import Audio

fname = "my_voice.wav"
if os.path.exists(fname):
    fs, x = wavfile.read(fname)
    if x.ndim > 1:
        x = x.mean(axis=1)
    x = x / np.max(np.abs(x))
else:
    fs = 16000
    t = np.arange(0, 2, 1 / fs)
    x = 0.5 * np.sin(2 * np.pi * 440 * t) * np.exp(-t)

t = np.arange(len(x)) / fs
print(f"fs = {fs} Hz, duration = {len(x)/fs:.2f} s, samples = {len(x)}")

plt.plot(t, x)
plt.xlabel("time (s)")
plt.ylabel("amplitude")
plt.title("My recording")
plt.show()

Audio(x, rate=fs)
```

```python
## Analysis questions

1. **Sampling rate & 1-minute mono storage:**  
   My recording's sampling rate is `fs = 16000 Hz` *(update with your phone's printed `fs` if using `my_voice.wav`)*.  
   For 1 minute (60 seconds) in mono, the number of samples needed is:  
   `fs × 60 = 16000 × 60 = 960,000` numbers *(update this product to match your `fs`)*.

2. **Why a stem plot is more appropriate than a continuous line for $x(n)$:**  
   A discrete-time signal $x(n)$ is only defined at integer sample indices $n$. Drawing a continuous line falsely implies that signal values exist between integers, whereas a stem plot accurately shows values only at integer points.

3. **Property violated by $a = 1.1$:**  
   For $a = 1.1$, $(1.1)^n u(n)$ grows without bound as $n \to \infty$. If this were a system's impulse response $h(n)$, it would violate **BIBO (Bounded-Input, Bounded-Output) stability**, because $\sum_{n=0}^{\infty} |h(n)| = \infty$.
```
